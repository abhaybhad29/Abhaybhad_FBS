# Merge

```python
emp = pd.DataFrame({'id':[101,102,103,104],'name':['Akshay','Rohit','Rahul','Vedant'],'did':[1,2,1,4]})
emp
```

```python
dept = pd.DataFrame({'did':[1,2,3],'dname':['Admine','IT','sales']})
dept
```

```python
pd.merge(emp,dept,how='inner')
```

```python
pd.merge(emp,dept,how='left')
```

```python
pd.merge(emp,dept,how="right")
```
