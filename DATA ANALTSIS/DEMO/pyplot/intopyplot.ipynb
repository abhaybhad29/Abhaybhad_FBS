# Bar Plot
1. Compairing categories
2. Showing frequencies or count

```python
product = ['Hot Coffee','Clod Coffee','Tea','Salad']
summer= [30,90,80,50]
```

```python
plt.figure(figsize=(7,7)) #(width,height)
plt.bar(product ,summer , color ='Orange')
plt.title('Sales of product for summer')
plt.xlabel('Products')
plt.ylabel('Count')
plt.show()
         
           
```
